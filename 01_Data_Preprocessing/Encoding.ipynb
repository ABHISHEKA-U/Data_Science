### Label Encoding :
It is encoding preprocessing technique which accepts one column and encodes the values from zero.

* It mainly used to encode the label(target) column

```python
# Encoding the target column using LabelEncoder

import pandas as pd

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Female", "Male", "Female"],
    "Purchased": ["Yes", "No", "Yes", "Yes", "No"]
})

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])
df["Purchased"] = le.fit_transform(df["Purchased"])
print(df)
```

### OneHot Encoding

Used in Linear Regression, Logistic Regression, Neural Nets

Nominal categories (color, city, brand, country) → OneHotEncoder
Ordered categories (small < medium < large, low < medium < high) → OrdinalEncoder
Tree models (Random Forest, XGBoost, LightGBM) can often work reasonably with OrdinalEncoder, even for nominal variables, though OneHotEncoder is still conceptually cleaner for unordered categories.

OneHotEncoder is used when the data is nominal and has less unique category,as one hot encoder creates n new columns for n unique category in a single column.

```python
df = pd.DataFrame({
    "City": ["Bangalore", "Mumbai", "Delhi", "Bangalore", "Delhi"],
    "Department": ["IT", "HR", "Finance", "IT", "HR"],
    "Salary": [50000, 40000, 60000, 55000, 45000]
})
df
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoded = encoder.fit_transform(
    df[["City", "Department"]]
)

print(encoded)
```

### Ordinal Encoding

Categories have natural order (Low, Medium, High)

Used in Tree models (XGBoost, LightGBM, Random Forest)

```python
df = pd.DataFrame({
    "Education": [
        "High School",
        "Bachelor's",
        "Master's",
        "PhD",
        "Bachelor's"
    ]
})
df
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[[
        "High School",
        "Bachelor's",
        "Master's",
        "PhD"
    ]]
)

df["Education_encoded"] = encoder.fit_transform(
    df[["Education"]]
)
print(df)
```

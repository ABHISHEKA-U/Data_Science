#####  PRECISION SCORE FOR +VE LABEL

Out of all the predicted +ve labels, how much % labels are correctly predicted as +ve.

```python
PRECISION(+ve) = TP / TP +FP

            # EX 1:
            #                                 100         200
            #                                 500         200

                        PRECISION(+ve) =  200 / (200+200)

            # # CONCLUSION 1:
            # out of all the predicted positive labels ( 400 ), 50% (TP) samples are correctly predicted as positive  , REMAINING % ARE FALSLY PREDICTED AS POSITIVE NOT NEGATIVE


            # EX 2:
            #                         80      20
            #                         100     300


            # # CONCLUSION 2:
            # out of all the predicted spam labels ( 320 ), 90% samples are correctly predicted as spam
```

### F1 SCORE [ USED FOR IMBALANCE DATASET ]
 
It is a classification metric which balances both precision and recall score and we use this for imbalance dataset.

```python
# F1 SCORE FOR NEGATIVE:

#                       = 2 * [ precision_negative * recall_neg ] / precision_negative + recall_negative


# F1 SCORE FOR POSITIVE :

#                       = 2 * [ precision_positive  * recall_positive  ] / precision_positive + recall_positive
```

#### ACCURACY SCORE ( for over all conclusion ) = USED FOR BALANCED DATASET

It is a classification metric which tells out of all the samples, how many samples are correctly predicted .

Accuracy score = (TN + TP )/ TN + FN + TP + FP

When to use ?

If class labels are balanced we can use accuracy score 

When not to use ?

If class labels are imbalance we cannot use accuracy score , if we use we will get wrong conclusion

### Imputation

* It is a preprocessing technique which handles missing values with the help of strategy parameter.

strategy : mean  |  median  |  most_frequent  | constant

For numerical column : mean and median
For categorical column : most frequent

constant - numerical : 0
           categorical: value

while using the constant strategy we should provide fill_value

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, 30, None, 45, 28, np.nan, 35, 40, 32, 27,25, 30, np.nan,28, np.nan, 35,30, np.nan,28, np.nan],
    "Salary": [40000, np.nan, 50000, 65000, 42000, 48000, np.nan, 70000, 52000, 45000,40000, np.nan, 50000,48000, np.nan, 70000, 52000,45000,40000, np.nan],
    "Department": [
        "HR",
        "IT",
        "Finance",
        np.nan,
        "IT",
        "HR",
        "Finance",
        np.nan,
        "IT",
        "HR",
        np.nan,
        "IT",
        "HR",
        "Finance",
        np.nan,
        "IT",
        "HR",
        "Finance",
        np.nan,
        "IT",
    ]
})

print(df)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
```

```python
impute = SimpleImputer(strategy='constant',fill_value='unknown')
impute.fit_transform(df)
```

```python
impute = SimpleImputer(strategy='mean')
impute.fit_transform(df[['Age','Salary']])
```

```python
impute = SimpleImputer(strategy='median')
impute.fit_transform(df[['Age','Salary']])
```

```python
impute = SimpleImputer(strategy='most_frequent')
impute.fit_transform(df[['Department']])
```

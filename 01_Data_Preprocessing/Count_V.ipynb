```python
import pandas as pd
import numpy as np
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer(stop_words='english')
```

```python
corpus = [
'It is a good movie',

'Family friendly movie',

'More into emotional',

'first half was good',

'Expected good ending']
```

```python
X = vectorizer.fit_transform(corpus)
```

```python
X
```

```python
print(X.toarray)
```

```python
val = X.toarray()
```

```python
cols = vectorizer.get_feature_names_out()
cols
```

```python
pd.DataFrame(data=val,columns=cols)
```

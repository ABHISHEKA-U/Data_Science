### PCA - Principal Component Analysis

It is a preprocessing technique which comes under dimensionality reduction, It reduces the dimension by using variance.

* We use this for two scenarios :

    1. Pre-processing technique
    
    2. 2 - D  Visualization

```python
# SYNTAX

# from sklearn.decomposition import PCA

# pca_2d = PCA(n_components=2)
# pca_2d.fit_transform(df)
```

* PCA does not remove the columns but reduces the dimensionality to n_components using the existing features.

When to use :

* when multi co-linearity problem occurs

* When model became over-fit( high variance) and it reduces the complexity.

* 2-D Visuals(ploting cluster values/labels)

* When we have high dimension(more features)

* Along with scaling techniques we use PCA for distance based algorithms(Preprocessing -> PCA -> Algorithm)

When not to use :

* For categorical columns after encoding

## Feature Scaling

While doing datapreprocessing feature scaling is neccessary to give equal importance to all the numerical columns. When we working the machine learning algorithms which uses distance(like euclidean distance) to calculate the dependent value, if the variable like p is high and q is low then the distance is influeced by the higher variable,eventually model will also. Then we loose the valuable information of the lower value variable, To address this we need to scale the features.

Scaling is a Preprocessing Technique which scales the values equally for all the columns.

By which we can give equal importance to all the numerical column in Distance based algorithm(linear regression,logistic Regression,KNN,SVM,K-means Clustering).

Scaling techniques are not used for tree based algorithms as they dont use any distance metric

Scaling is applied only on numerical columns not for object/categorical columns.

When numerical column values are in different range then we use scaling techniques.

scaling is done on features column,not on target column.

```python
# Scaling
# |
# |- Standardization : Scales the values Not in fixed range | Use it when data have outliers(find using
# |  boxplot)
# |  |-Standard Scaling : Scales the values between no range. |use when we have less outliers
# |  | Uses the z-score = (x-mean)/sd.| to check after scaling the mean~0, sd~1
# |  |
# |  |-Robust Scaling : Scales the values between no range | use when we have more outliers
# |  | x-Q(median)/IQR(Q3-Q1)
# |
# |- Normalization : Scales the values in fixed range(0-1) |Use it when data is clean
#    |- Min-Max Scaling
#      x-xmin/xmax-xmin
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('icecream.csv').dropna()
```

```python
X = df[[ "Temperature (°C)"]]
Y = df[[ "Ice Cream Sales (units)"]]
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler,RobustScaler,MinMaxScaler,Normalizer
```

```python
standard_scaler = StandardScaler()
Robust_scaler = RobustScaler()
min_max_scaler = MinMaxScaler()

print("Standardization")
print(standard_scaler.fit_transform(X))
```

```python
print("Robust Scaling")
print(Robust_scaler.fit_transform(X))
```

```python
print("MinMax Scaling")
print(min_max_scaler.fit_transform(X))
```

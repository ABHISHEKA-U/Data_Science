### Naive Bayes

It is a supervised learning algorithm which is used for classification scenarios and it uses probability to predict the class labels based on highest possibility.

Where we use this :

Classification Scenarios like

    * Sentiment Analysis
    * Medical Diagnosis


Algorithms Naive Bayes Use :

>> Bernoulli Naive Bayes - Used when feature column are binary(classification)

>> Gaussian Naive Bayes - Used When feature column values are continuos(classification)

>> Multi-nominal Naive Bayes - Used for sentence based classification

>>>> Logistic Regression will be the first choice for sentence based classification(like sentiment Analysis ) and second would be Naive Bayes.(Both  well suited for sentence based classification because both use probability)

Multi-nominal Naive bayes is the fastest supervised learning algorithm.

Similar to KNN it also just stores the data and make prediction while executing

### LOGISTIC REGRESSION

It is the supervised learning algorithm which comes under regression and used for classification scenario

It follows S curve pattern

What is classification algorithms ?

    it is a supervised learning technique which predicts categorical values

example :

        * spam detection [ here the classification algorithm predicts spam and not spam labels ]

        * sentiment analysis [ predicts +ve,-ve,neutral]




how to select algorithms(Regression/Classification) ?

    * if target column values are continues then regression algorithms
    
    * if target column values are categorical then we use classification algorithms

#### it is for classification but why it is named as regression ?

> logistic means it uses sigmoid function

* sigmoid function mostly we use for binary class labels ( target column values ) 
* sigmoid classifies the binary labels(0 or 1)
* only 2 labels it will work not more then that.

it follows S curve pattern

it internally uses
                    1 / 1 + e^-z

here z means it uses regression formula internally => w1x1+w2x2+w3x3

here wights are not slopes

we use 1 in numerator because the value will be in  between zero and one

[ >=0.5 ->1 , <0.5 -> 1 (threshold O.5 DEFAULT)]

#### WHY NOT LINEAR REGRESSION FOR BINARY CLASSIFICATION

In linear regression it find straight line (continues prediction)  but here the target column we gave 0 , 1 already, then how it will predict continuous values.

* In linear regression if x increase y will also increase ( y will always increase), it will predict continuos values , not between 0-1

* In logistic regression if x increase y will be in 0-1 only , it is fixed that it will predict the values between 0-1


therefore we will not use linear regression for classification scenario

HOW WEIGHTS ARE CALCULATED IN LOGISTIC REGRESSION 

In linear regression how it calculates the weights

    in linear Regression weights = slope (formula)


But here in logistic regression weights are not slopes

    internally it uses optimization technique called gradient descent


1. first randomly it initilize the weights between 0 and 0.5

2. and later with the help of gradient descent , it finds the best weights

3. to find or optimize the best weights , internally it takes multiple iterations


#### Techniques logistic regression uses internally

1.loss function

2.cost function

3.sigmoid function

4.regularization techniques (L1,L2,elastic net)

5.gradient descent


01 . LOSS FUNCTION

Loss function measures error per record( row wise one by one )

02 . COST FUNCTION

Cost function measures overall model mistakes (overall not one by one) by using  log Loss

03 . SIGMOID FUNCTION

To get binary value

04 . REGULARIZATION

When the model is over-fit we use regularization to reduce the variance

with the help of PCA and regularization we can reduce the variance

WHAT IS REGULARIZATION ?

    It is an ml technique which helps us to reduce the variance with the help of penalties (L1,L2,ELASTIC NET)

WHAT IS GENERALIZATION ?

    How well the model captures the pattern of unseen data

        * If model captures the pattern of unseen data then the model is generalized well

        * If model fails to capture the unseen data pattern then the model is not generalized well [ OVER-FIT]

Then we make use of Regularization

PENALTIES

To correct the model mistake we are adding penalties

* L1=lasso

* L2=ridge

* Elastic-Net = L1 + L2

It  measures the mistake by cost function

When we use logistic regression it internally uses the Loss,Cost,Sigmoid function , these calculates the model mistakes
to correct those mistake we use the regularization by adding the penalties like L1, L2, elastic-net

WHAT IS LASSO ? [L1] - pushes weights to zero

    it is a regularization penalty which selects the important features by pushing the weights towards to zero and exactly to zero.

        if the weight is zero which means, that feature is not important

HOW IT CHOOSE THE IMPORTANT FEATURE - by pushing the weights towards to zero

    In the linear regression scenario we have the concept called slope which means weight,

        weight means the importance of that particular feature

        consider example like slop m1 = 0 and m2=5 and c=10,x1=5,x2=6

        the equation for 2 features becomes y =  m1x1+m2x2+c

            x1 => y=0*5+5*6+10

    here the L1 will not choose the x1 , because the weight is zero that means feature x1 is not important
    
WHEN TO USE L1 ?

    * When we have more features
    * For sentence based classification
    * When we don't know the important features

with the help of Gradient Descent it pushes the weights towards zero

L1- When we have more features - lasso will select important features

L2  - We use when multi-colinearity problem occurs and when all features are important

Elastic-net - when we have more features and when multicolinearity

WHAT IS RIDGE ? [l2] - shrinks the weights

    It is a regularization penalty which keeps all the columns by shrinking the weights

    it shrinks the weights but never pushes to zero


WHEN TO USE ?
    When all the feature columns are important

    when multi-colinearity problem occurs

WHAT IS ELASTIC-NET:

    It is a regularization penalty which uses both l1 and l2

    it selects important features and also handles multi-colinearity problem

how elastic-net uses both l1 and l2 ?

    it handles both L1 and L2 with the help of L1 RATIO

        * L1 ratio = 0.3

        * L1 ratio = 0(0% for L1) == L2

        * L1 ratio =1(100% for L1) == L1


GRADIENT DESCENT 

WHAT IS GRADIENT DESCENT ?

    it is an optimization technique which is used to find the best weights

WHERE WE GD IN ML ?

    * Logistic Regression internally uses Gradient Descent to find the best weights

    * Extreme Gradient Boosting algorithm [ XG - BOOST ] uses Gradient Descent to find the best weights

    * Regularization techniques uses Gradient Descent to find the best weights and reduces the variance.( because linear regression also uses regression - not by default)

INTERNAL FLOW OF LOGISTIC REGRESSION

    1.Randomly initialize the weights ( no fixed formulas to find best weights )
        as it is initialize randomly , if we use that weights we get high errors , that we can solve by gradient descent.

    to measure the errors
    Sigmoid will return confidence score

        actual - confident score = error

    2. It uses cost function - those errors are measured by log loss ( overall mistake of model)

    3. Regularization - add penalties to reduce model mistake

    4. Gradient descent

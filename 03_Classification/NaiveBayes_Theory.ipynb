### Conditional Probability

If there are two events A and B then Probability A given Probability of B is written as :

P(A|B) = P(A ∩ B) / P(B)

given, P(B) <> 0

Example :

Rolling 2 dices(D1, D2)

Sample_Space = {(1,1),(1,2)(1,3),(1,4),(1,5),(1,6),
                (2,1),......................,(6,6)}

* P(A = 5)= 1/36

* P(B)=((D1 + D2 ) <= 10) = 33 / 36 = 11 / 12

* A = {D1 = 5} , B = {(D1 + D2) <= 10 }

    P(A | B) = P(A ∩ B) / P(B) ----> ( 5/36 ) / ( 33/36 ) = 5 / 36
    
    P(A ∩ B) = 5/36

    P(B) = 33/36


#### Independent Events

A and B are called Independent if :

P(A ∩ B) = P(A) * P(B) -----------(1)

P(A|B) = P(A ∩ B) / P(B)

P(A|B) = P(A) * P(B) / P(B)

P(A|B) = P(A) ------------(2)

CONCLUSION : If A and B are Independent Events the Probability of any(A / B) of the event given another occurred is solely depend on Probability of Event.


#### Mutual Exclusive Events

A and B are called Mutually Exclusive if :

P(A ∩ B) = 0  -----------(1)

P(A|B) = P(A ∩ B) / P(B)

P(A|B) = 0 -----------(2)

CONCLUSION : If A and B are Mutually Exclusive Events the Probability of any(A / B) of the event given another occurred is 0 .

Eg . The probability of getting tail given head is occurred is zero.

#### Bayes Theorem

Bayes theorem states that if there are two Events A and B,then

P(A | B) = ( P(B | A) P(A) ) / P(B)

given P(B) <> 0

Where :

    *   P(A | B) : Posterior Probability

    *   P(B | A) : Likely Hood Probability

    *   P(A)     : Prior Probability

    *   P(B)     : Evidence
